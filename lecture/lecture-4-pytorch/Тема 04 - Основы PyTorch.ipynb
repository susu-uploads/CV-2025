## Введение в тензоры 
Тензоры являются фундаментальным строительным блоком машинного обучения. Их работа заключается в представлении данных в числовом виде.

Например, вы можете представить изображение в виде тензора с формой `[3, 224, 224]`, что будет означать `[color_channels, height, width]`: `3` - количество цветовых канала (красный, зеленый, синий), высота 224 пикселя и ширина 224 пикселя. Тензор будет иметь три измерения, одно для «color_channels», второе для «height» и третье для «width».

### Создание тензоров

Есть целая страница документации, посвященная классу [`torch.Tensor`](https://pytorch.org/docs/stable/generated/torch.tensor.html#torch.tensor).

Первое, что мы создадим, это **scalar**.

Scalar — это одно число, и, говоря на тензорном языке, это тензор нулевой размерности.

````python
# Scalar
scalar = torch.tensor(7)
scalar
````

`scalar` является числом, но имеет тип `torch.Tensor`.

Мы можем проверить размеры тензора, используя атрибут `ndim`.

````python
scalar.ndim
````

Чтобы получить число из тензора, необходимо использовать метод `item()`. Это позволит получить число типа integer, а не `torch.Tensor`.

````python
# Получаем число (работает только с одноэлементными тензорами)
scalar.item()
````

Теперь рассмотрим следующее важное понятие в Torch, вектор. Вектор — это одномерный тензор, который может содержать множество чисел.

Например, у вас может быть вектор `[3, 2]` для описания `[спальни, ванные]` в вашем доме.

````python
# Vector
vector = torch.tensor([7, 7])
vector
````

Сейчас вектор содержть два числа 7. Что с измерениями?

````python
# Проверка количества измерений вектора
vector.ndim
````

Почему числа 2, а измерение 1? Количество измерений определяется по количеству квадратных скобок.

Иногда, измерения также называют рангом матрицы. Тензоры принято изображать в табличной форме: вектор (ndim=1) - это строка чисел, матрица формы (rows,cols) - это прямоугольная таблица с rows строчками и cols колонками.

Еще одним важным понятием для тензоров является их атрибут `shape` («форма»). Форма говорит, как расположены элементы внутри них.

Давайте проверим форму вектора.

````python
# Проверяем форму вектора
vector.shape
````

`torch.Size([2])`означает, что вектор имеет форму `[2]`. Это из-за двух элементов, которые мы поместили в квадратные скобки (`[7, 7]`).

Также, важным элементом в Torch является матрица.

````python
# Матрица
MATRIX = torch.tensor([[7, 8], 
                       [9, 10]])
MATRIX
````

Важно: у матриц есть дополнительное измерение.

````python
# Проверим количество измерений
MATRIX.ndim
````

Проверим форму:

````python
MATRIX.shape
````

Мы получаем вывод `torch.Size([2, 2])`, потому что `MATRIX` имеет два элемента в глубину и два элемента в ширину.

Создадим тензор:

````python
# Тензор
TENSOR = torch.tensor([[[1, 2, 3],
                        [3, 6, 9],
                        [2, 4, 5]]])
TENSOR
````

Тензоры могут представлять все что угодно.

````python
# Проверяем количество измерений
TENSOR.ndim
````

````python
# Проверяем форму
TENSOR.shape
````

Мы получили `torch.Size([1, 3, 3])`. Размеры идут от внешнего к внутреннему, что означает, что есть 1 измерение 3 на 3.

![example of different tensor dimensions](https://raw.githubusercontent.com/mrdbourke/pytorch-deep-learning/main/images/00-pytorch-different-tensor-dimensions.png)

Использование строчных букв для `scalar` и `vector` и заглавные буквы для `MATRIX` и `TENSOR`. Это было специально. На практике вы часто будете встречать скаляры и векторы, обозначаемые строчными буквами, такими как `y` или `a`. И матрицы, и тензоры обозначаются прописными буквами, такими как `X` или `W`.

Let's summarise.

| Name | What is it? | Number of dimensions | Lower or upper (usually/example) |
| ----- | ----- | ----- | ----- |
| **scalar** | одно число | 0 | Lower (`a`) | 
| **vector** | число или много числе | 1 | Lower (`y`) |
| **matrix** | n-мерный массив чисел | 2 | Upper (`Q`) |
| **tensor** | n-мерный массив чисел | может быть любым числом, 0-мерный тензор является скаляром, 1-мерный тензор является вектором | Upper (`X`) |

### Случайные тензоры

Тензоры представляют некоторую форму данных. Модели машинного обучения манипулируют тензорами и ищут закономерности. Однако, при создании моделей машинного обучения с помощью PyTorch вы редко будете создавать теноров вручную. 
Вместо этого модель машинного обучения часто начинает со случайных тензорных чисел и корректирует эти случайные числа по мере обработки данных, чтобы лучше их представлять. По сути: «Начать со случайных чисел -> просмотреть данные -> обновить случайные числа -> просмотреть данные -> обновить случайные числа...»

Как специалист по данным, вы можете определить, как модель машинного обучения запускается (инициализация), просматривает данные (представление) и обновляет (оптимизирует) свои случайные числа.

Давайте посмотрим, как создать тензор случайных чисел. Мы можем сделать это, используя метод [`torch.rand()`](https://pytorch.org/docs/stable/generated/torch.rand.html), передав параметр `size`.

````python
# Создадим случайный тензор размера (3, 4)
random_tensor = torch.rand(size=(3, 4))
random_tensor, random_tensor.dtype
````

Гибкость `torch.rand()` заключается в том, что мы можем настроить размер так, как захотим.

Например, предположим, что нам нужен случайный тензор в общей форме изображения `[224, 224, 3]` (`[height, width, color_channels`]).

````python
# Создадим случайный тензор размера (224, 224, 3)
random_image_size_tensor = torch.rand(size=(224, 224, 3))
random_image_size_tensor.shape, random_image_size_tensor.ndim
````

### Создание диапазона и тензоров
Иногда вам может понадобиться диапазон чисел, например, от 1 до 10 или от 0 до 100. Для этого можно использовать `torch.arange(start, end, step)`.

Where:
* `start` = начало диапазона (включая 0)
* `end` = конец диапазона (не включая 10)
* `step` = сколько шагов между каждым значением (1)

В Python вы можете использовать `range()` для создания диапазона. Однако в PyTorch `torch.range()` устарел и может отображать ошибку в будущем.

````python
# Используем torch.range() 
zero_to_ten_deprecated = torch.range(0, 10)
````

````python
# Используем torch.arange()
zero_to_ten = torch.arange(start=0, end=10, step=1)
zero_to_ten
````

Иногда вам может понадобиться один тензор определенного типа с той же формой, что и другой тензор. Например, тензор из нулей той же формы, что и предыдущий тензор. Для этого вы можете использовать метод [`torch.zeros_like(input)`](https://pytorch.org/docs/stable/generated/torch.zeros_like.html) или [`torch.ones_like(input)`](https://pytorch.org/docs/1.9.1/generated/torch.ones_like.html), которые возвращают тензор, заполненный нулями или единицами той же формы, что и предыдущий.

````python
# Создание тензора из нулей нужной формы
zero_to_ten = torch.arange(start=0, end=11, step=1)
ten_zeros = torch.zeros_like(input=zero_to_ten) # взятие нужной формы тензора
ten_zeros
````

### Тензорные типы данных

В PyTorch доступно множество различных [тензорных типов данных](https://pytorch.org/docs/stable/tensors.html#data-types). Некоторые из них для процессора, а некоторые для графического процессора. Как правило, если вы где-нибудь видите torch.cuda, тензор используется для графического процессора (поскольку графические процессоры Nvidia используют набор вычислительных инструментов под названием CUDA). Наиболее распространенный тип (и обычно используемый по умолчанию) — `torch.float32` или `torch.float`. По другому это называется 32-битной плавающей точкой. Есть также 16-битные числа с плавающей запятой (`torch.float16` или `torch.half`) и 64-битные числа с плавающей запятой (`torch.float64` или `torch.double`) и 8-битные, 16-битные, 32-битные и 64-битные целые числа.

Зачем так много и почему по умолчанию тип с плавающей точкой. Потому что в вычислениях важна точность. Точность в Torch - это количество цифр, используемых для описания числа. Чем выше значение точности (8, 16, 32), тем больше цифр и, следовательно, больше данных используется для выражения числа. Типы данных с более низкой точностью, как правило, быстрее вычисляются, но жертвуют некоторой производительностью по таким показателям оценки, например точностю (быстрее вычисляются, но менее точны).

На почитать: 
  * [PyTorch documentation for a list of all available tensor datatypes](https://pytorch.org/docs/stable/tensors.html#data-types).

Создадим некоторые тензоры с определенными типами данных. Мы можем сделать это, используя параметр `dtype`.

````python
# Тип данных по умолчанию для тензоров — float32.
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None, # по умолчанию — None, то есть torch.float32 или любой другой переданный тип данных
                               device=None, # по умолчанию None, который использует тип тензора по умолчанию
                               requires_grad=False) # если True, операции, выполняемые над тензором, записываются

float_32_tensor.shape, float_32_tensor.dtype, float_32_tensor.device
````

Помимо проблем с формой (тензорные формы не совпадают), две другие наиболее распространенные проблемы, с которыми вы столкнетесь в PyTorch, — это проблемы с типом данных и устройством.

Например, один из тензоров — `torch.float32`, а другой — `torch.float16` (в PyTorch важно, чтобы тензоры были одного формата).

Или один из ваших тензоров находится на процессоре, а другой — на графическом процессоре (в PyTorch также важно, чтобы вычисления между тензорами выполнялись на одном устройстве).

Давайте создадим тензор с `dtype=torch.float16`.

````python
float_16_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=torch.float16)

float_16_tensor.dtype
````

## Манипулирование тензорами (тензорные операции)

В глубоком обучении данные (изображения, текст, видео, аудио и т. д.) представляются в виде тензоров. Модель обучается, исследуя эти тензоры и выполняя серию операций (может быть 1 000 000+) над тензорами, чтобы создать представление шаблонов во входных данных. 

Примеры основных операций:
* Добавление
* Вычитание
* Умножение (поэлементно)
* Деление
* Матричное умножение

### Основные операции

Рассмотрим сложение (`+`), вычитание (`-`), умножение (`*`).

````python
# Создадим тензор значений и добавим к нему число
tensor = torch.tensor([1, 2, 3])
tensor + 10
````

````python
# Умножение на 10
tensor * 10
````

Приведенные выше значения тензора не оказались равны `tensor([110, 120, 130])`, это потому, что значения внутри тензора не меняются, если они не переназначены.

````python
# Тензоры не меняются, если их не переназначить
tensor
````

Давайте вычтем число, и на этот раз мы переназначим переменную `tensor`.

````python
# Вычетание и переназначение
tensor = tensor - 10
tensor
````

````python
# Добавление и переназначение
tensor = tensor + 10
tensor
````

PyTorch также имеет множество встроенных функций, таких как [`torch.mul()`](https://pytorch.org/docs/stable/generated/torch.mul.html#torch.mul) (сокращение от умножения) и [`torch.add()`](https://pytorch.org/docs/stable/generated/torch.add.html) для выполнения основных операций.

````python
# Использование встроенных функций
torch.multiply(tensor, 10)
````

````python
# Исходный тензор остается неизменным
tensor
````

Чаще используются операторные символы, такие как `*` вместо `torch.mul()`.

````python
# Поэлементное умножение (каждый элемент умножает свой эквивалент, индекс 0->0, 1->1, 2->2)
print(tensor, "*", tensor)
print("Equals:", tensor * tensor)
````

### Умножение матриц

Одной из наиболее распространенных операций в алгоритмах машинного обучения это [умножение матриц] (https://www.mathsisfun.com/алгебра/matrix-multiplying.html).

В PyTorch реализуется умножения матриц в методе [`torch.matmul()`](https://pytorch.org/docs/stable/generated/torch.matmul.html).

Следует помнить два основных правила умножения матриц:
1. Соответствие размеров матриц:
  * `(3, 2) @ (3, 2)` не будет работать
  * `(2, 3) @ (3, 2)` будет работать
  * `(3, 2) @ (2, 3)` будет работать
2. Полученная матрица имеет форму внешних размеров:
 * `(2, 3) @ (3, 2)` -> `(2, 2)`
 * `(3, 2) @ (2, 3)` -> `(3, 3)`

Все правила умножения матриц, используя метод `torch.matmul()` описаны в документации: [in the PyTorch documentation](https://pytorch.org/docs/stable/generated/torch.matmul.html).

Создадим тензор и выполним над ним поэлементное умножение и матричное умножение.

````python
import torch
tensor = torch.tensor([1, 2, 3])
tensor.shape
````

Разница между поэлементным умножением и матричным умножением заключается в сложении значений.

Для нашей переменной `tensor` со значениями `[1, 2, 3]`:

| Операция | Расчет | Код |
| ----- | ----- | ----- |
| Поэлементное умножение | `[1*1, 2*2, 3*3]` = `[1, 4, 9]` | `tensor * tensor` |
| Умножение матриц | `[1*1 + 2*2 + 3*3]` = `[14]` | `tensor.matmul(tensor)` |


````python
# Поэлементное матричное умножение
tensor * tensor
````

````python
# Умножение матриц
torch.matmul(tensor, tensor)
````

````python
# Также можно использовать символ «@» для умножения матриц, хотя это не рекомендуется
tensor @ tensor
````

Вы можете выполнять умножение матриц вручную, но это не рекомендуется. Встроенный метод `torch.matmul()` работает быстрее.

````python
%%time
# Умножение матриц вручную 
# необходимо избегайть выполнения операций с циклами for, они требуют больших вычислительных ресурсов
value = 0
for i in range(len(tensor)):
    value += tensor[i] * tensor[i]
value
````

````python
%%time
torch.matmul(tensor, tensor)
````

## Одна из самых распространенных ошибок в глубоком обучении (ошибки формы)

Поскольку большая часть глубокого обучения заключается в умножении и выполнении операций над матрицами, а матрицы имеют строгое правило о том, какие формы и размеры можно комбинировать, одна из наиболее распространенных ошибок, с которыми вы столкнетесь — это несоответствие форм.

````python
# Формы должны быть правильными  
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)

torch.matmul(tensor_A, tensor_B) # (это будет ошибкой)
````

Мы можем выполнить матричное умножение между `tensor_A` и tensor_B, совместив их внутренние размеры.

Один из способов сделать это — транспонировать (переключить размеры данного тензора).

Вы можете выполнять транспонирование в PyTorch, используя:
* `torch.transpose(input, dim0, dim1)` - где `input` — желаемый тензор для транспонирования, а `dim0` и `dim1` — размеры, которые нужно поменять местами.
* `tensor.T` - где `tensor` - желаемый тензор для транспонирования.

````python
# Посмотрим тензор_А и тензор_В
print(tensor_A)
print(tensor_B)
````

````python
# Просмотр tensor_A и tensor_B.T
print(tensor_A)
print(tensor_B.T)
````

````python
# Операция сработает при транспонировании tensor_B
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}\n")
print(f"New shapes: tensor_A = {tensor_A.shape} (same as above), tensor_B.T = {tensor_B.T.shape}\n")
print(f"Multiplying: {tensor_A.shape} * {tensor_B.T.shape} <- inner dimensions match\n")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output) 
print(f"\nOutput shape: {output.shape}")
````

Вы также можете использовать [`torch.mm()`](https://pytorch.org/docs/stable/generated/torch.mm.html), который является сокращением от `torch.matmul()`.

Нейронные сети используют матричные умножения и скалярные произведения.

Модуль [`torch.nn.Linear()`](https://pytorch.org/docs/1.9.1/generated/torch.nn.Linear.html), также известный как линейный слой реализует матричное умножение между входным значением «x» и матрицей весов «A».

$$
y = x\cdot{A^T} + b
$$

Где:
* `x` — это входные данные для слоя.
* `A` — это матрица весов, созданная слоем, она начинается со случайных чисел, которые корректируются по мере того, как нейронная сеть учится лучше представлять шаблоны в данных (`T`, так как матрица весов транспонируется).
Вы также можете часто видеть «W» или другую букву, например «X», используемую для демонстрации матрицы весов.
* `b` - это термин смещения, используемый для небольшого смещения весов и входных данных.
* `y` - является выходом (манипулирование входом в надежде обнаружить в нем закономерности).

`torch.nn.Linear()` - это линейная функция.

````python
# Поскольку линейный слой начинается с матрицы случайных весов, давайте сделаем его воспроизводимым
torch.manual_seed(42)
# Матричное умножение
linear = torch.nn.Linear(in_features=2, # in_features = соответствует внутреннему размеру ввода 
                         out_features=6) # out_features = описывает внешнее значение 
x = tensor_A
output = linear(x)
print(f"Input shape: {x.shape}\n")
print(f"Output:\n{output}\n\nOutput shape: {output.shape}")
````

### Reshaping, stacking, squeezing and unsqueezing

Часто вам нужно изменить форму или размеры ваших тензоров, фактически не изменяя значения внутри них.

Для этого есть несколько популярных способов:

| Метод | Описание |
| ----- | ----- |
| [`torch.reshape(input, shape)`](https://pytorch.org/docs/stable/generated/torch.reshape.html#torch.reshape) | Преобразовывает `input` в `shape` (если совместимо), можно использовать метод `torch.Tensor.reshape()`. |
| [`torch.Tensor.view(shape)`](https://pytorch.org/docs/stable/generated/torch.Tensor.view.html) | Возвращает представление исходного тензора в другой `shape`, но использует те же данные, что и исходный тензор. |
| [`torch.stack(tensors, dim=0)`](https://pytorch.org/docs/1.9.1/generated/torch.stack.html) | Объединяет последовательность `tensors` в новом измерении (`dim`), все тензоры должны быть одного размера. |
| [`torch.squeeze(input)`](https://pytorch.org/docs/stable/generated/torch.squeeze.html) | Сжимает `input`, чтобы удалить все измерения со значением `1`. |
| [`torch.unsqueeze(input, dim)`](https://pytorch.org/docs/1.9.1/generated/torch.unsqueeze.html) | Возвращает `input` со значением измерения `1`, добавленным в `dim`. | 
| [`torch.permute(input, dims)`](https://pytorch.org/docs/stable/generated/torch.permute.html) | Возвращает представление исходного `input` с его размерами, переставленными (переставленными) в `dims` | 

Для чего это? Модели машинного обучения каким-то образом манипулируют тензорами. И из-за правил умножения матриц, если у вас есть несоответствие формы, вы столкнетесь с ошибками. Эти методы помогут сделать так, чтобы правильные элементы тензоров смешивались с правильными элементами других тензоров.

````python
# Создаем тензор
import torch
x = torch.arange(1., 8.)
x, x.shape
````

Теперь давайте добавим дополнительное измерение с помощью `torch.reshape()`.

````python
# Добавляем дополнительное измерение
x_reshaped = x.reshape(1, 7)
x_reshaped, x_reshaped.shape
````

Мы также можем изменить вид с помощью `torch.view()`.

````python
# Изменить представление (сохраняются те же данные, что и исходные, но изменяется вид)
z = x.view(1, 7)
z, z.shape
````

Однако помните, изменение вида тензора с помощью `torch.view()` на самом деле создает только новый вид того же тензора. Таким образом, изменение вида также меняет исходный тензор.

````python
# Изменение z изменяет x
z[:, 0] = 5
z, x
````

Если бы мы хотели сложить наш новый тензор поверх самого себя пять раз, мы можем сделать это с помощью `torch.stack()`.

````python
# Складываем тензоры друг на друга
x_stacked = torch.stack([x, x, x, x], dim=0) # попробуй изменить dim на dim=1 и посмотри что будет
x_stacked
````

Еще может понадобится удаление всех одиночных измерений из тензора. Для этого можно использовать `torch.squeeze()`.

````python
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimension from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
````

И чтобы сделать обратное `torch.squeeze()`, можно использовать `torch.unsqueeze()`, чтобы добавить значение измерения 1 к определенному индексу.

````python
print(f"Previous tensor: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

## Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
````

Можно изменить порядок значений осей с помощью `torch.permute(input, dims)`, где `input` превращается в вид с новыми `dims`.

````python
# Создаем тензор с определенной формой
x_original = torch.rand(size=(224, 224, 3))

# Переставляем исходный тензор, чтобы изменить порядок оси
x_permuted = x_original.permute(2, 0, 1) # смещает ось 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
````

Поскольку перестановка возвращает представление (совместно использует те же данные, что и оригинал), значения в переставленном тензоре будут такими же, как исходный тензор, и если вы измените значения в представлении, оно изменит значения оригинала.

````python
# Тензор в массив NumPy
tensor = torch.ones(7) # Создаем тензор единиц с dtype=float32
numpy_tensor = tensor.numpy() # будет dtype=float32, если не будет изменено
tensor, numpy_tensor
````

## Воспроизводимость (попытка исключить случайное из случайного)

Как случайность связана с нейронными сетями и глубоким обучением?

Нейронные сети начинают со случайных чисел для описания закономерностей в данных и пытаются улучшить эти случайные числа, используя тензорные операции, чтобы лучше описывать закономерности в данных.

Иногда необходимо, что бы случайность была закономерна. Мы можем проводить повторяемые эксперименты. Например, вы создаете алгоритм, способный достичь производительности X. Другими словами, можете ли вы получить такие же (или очень похожие) результаты на разных компьютерах. Давайте посмотрим на краткий пример воспроизводимости в PyTorch. Мы начнем с создания двух случайных тензоров, поскольку они случайны, мы ожидаем, что они будут разными.

````python
import torch

# Создаем два случайных тензора
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(f"Tensor A:\n{random_tensor_A}\n")
print(f"Tensor B:\n{random_tensor_B}\n")
print(f"Does Tensor A equal Tensor B? (anywhere)")
random_tensor_A == random_tensor_B
````

Как и следовало ожидать, тензоры получаются с разными значениями. Но что, если вы хотите создать два случайных тензора с одинаковыми значениями. Например, тензоры по-прежнему будут содержать случайные значения, но будут одинаковыми.

````python
import torch
import random

# Установите случайное начальное число
RANDOM_SEED=42 # попробуйте изменить это на другие значения и посмотреть, что произойдет с числами ниже
torch.manual_seed(seed=RANDOM_SEED) 
random_tensor_C = torch.rand(3, 4)

# Приходится сбрасывать начальное значение каждый раз, когда вызывается новый rand() 
# Без этого tensor_D будет отличаться от tensor_C. 
torch.random.manual_seed(seed=RANDOM_SEED) # попробуйте закомментировать эту строку и посмотреть, что произойдет
random_tensor_D = torch.rand(3, 4)

print(f"Tensor C:\n{random_tensor_C}\n")
print(f"Tensor D:\n{random_tensor_D}\n")
print(f"Does Tensor C equal Tensor D? (anywhere)")
random_tensor_C == random_tensor_D
````

То, что мы только что рассмотрели, лишь поверхностно касается воспроизводимости в PyTorch. Чтобы узнать больше о воспроизводимости в целом и случайных семенах:
* [Документация по воспроизводимости PyTorch](https://pytorch.org/docs/stable/notes/randomness.html).
* [Страница случайных начальных чисел Википедии](https://en.wikipedia.org/wiki/Random_seed).

### 3. Помещение тензоров (и моделей) на GPU

Вы можете поместить тензоры (и модел) на конкретное устройство, вызвав [`to(device)`](https://pytorch.org/docs/stable/generated/torch.Tensor.to.html) на них. `device` — это целевое устройство, на которое вы хотите передать тензор (или модель).

Помещение тензора на GPU с помощью to(device) (например, some_tensor.to(device)) возвращает копию этого тензора, например один и тот же тензор будет на CPU и GPU. Чтобы перезаписать тензоры, необходимо переназначить их:

some_tensor = some_tensor.to(устройство)

Попробуем создать тензор и закинуть его на GPU (если он есть).

````python
# Создаем тензор (по умолчанию на CPU)
tensor = torch.tensor([1, 2, 3])

# Тензор не на GPU
print(tensor, tensor.device)

# Переместим тензор в GPU (если доступно)
tensor_on_gpu = tensor.to(device)
tensor_on_gpu
````

Если у вас есть доступный графический процессор, приведенный выше код выведет:

```
tensor([1, 2, 3]) cpu
tensor([1, 2, 3], device='cuda:0')
```

Обратите внимание, что второй тензор имеет `device='cuda:0'`, это означает, что он хранится на 0-м доступном графическом процессоре (графические процессоры имеют индекс 0, если бы были доступны два графических процессора, они были бы `'cuda:0'` и `' cuda:1'` соответственно, до `'cuda:n'`).

### 4. Перемещение тензоров обратно в CPU
Зачем это делать? Например, мы хотим взаимодействовать со своими тензорами с помощью NumPy (NumPy не использует графический процессор).

Давайте попробуем использовать метод [`torch.Tensor.numpy()`](https://pytorch.org/docs/stable/generated/torch.Tensor.numpy.html) на нашем `tensor_on_gpu`.

````python
# Если тензор находится на графическом процессоре, его нельзя преобразовать в NumPy (это приведет к ошибке)
tensor_on_gpu.numpy()
````

Вместо этого, чтобы вернуть тензор процессору и использовать его с NumPy, мы можем использовать [`Tensor.cpu()`](https://pytorch.org/docs/stable/generated/torch.Tensor.cpu.html).

Это копирует тензор в память CPU.
